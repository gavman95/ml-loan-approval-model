```python
df.columns = [c.strip() for c in df.columns]
```

```python
df['loan_status'].value_counts()
```

```python
X = df.drop(columns=['loan_status'])
y = df['loan_status']



```

```python
X.info()
```

```python
# Apply domain knowledge as part of feature engineering

X['total_assets_value'] = X['residential_assets_value'] + X['commercial_assets_value'] + X['luxury_assets_value'] + X['bank_asset_value']

X.drop(columns=['residential_assets_value', 'commercial_assets_value', 'luxury_assets_value', 'bank_asset_value'], inplace=True)
```

```python
X['education'].unique()
```

```python
X['self_employed'].unique()
```

```python
y.unique()
```

```python
X['education'] = X['education'].str.strip()
X['self_employed'] = X['self_employed'].str.strip()
y = y.str.strip()
```

```python
X['education'].unique()
```

```python
# Convert columns into categorical variables
X['education'] = np.where(X['education'] == 'Graduate', 1, 0)
X['self_employed'] = np.where(X['self_employed'] == 'Yes', 1, 0)
```

```python
X['education'].unique()
```

```python
X.head()
```

```python
y.head()
```

```python
y = np.where(y == 'Approved', 1, 0)
```

```python
y
```

```python
# Log transformation
log_cols = ['income_annum', 'loan_amount', 'total_assets_value']
X[log_cols] = np.log(X[log_cols])
```

```python
X.head()
```

```python
X.drop(columns=['loan_id'], inplace=True)
```

```python
# Building the model
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)


```

```python
lr = LogisticRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)



```

```python
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc}")
```
